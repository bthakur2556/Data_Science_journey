## 1. Probability mass function(PMF):

**a. PMF of rolling a dice:**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import random
L = []                        #empty list jisme following result store krna hai
for i in range(1000):
    L.append(random.randint(1,6))     
```

```python
L[:5]     #first five outcomes
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()     #hamne un sabki probability nikaldi
```

```python
s
```

```python
s.plot(kind= "bar")
```

**b. Rolling two dices simultaneously:**

```python
L = []                        #empty list jisme following result store krna hai
for i in range(1000):
    a = random.randint(1,6) 
    b = random.randint(1,6) 
    L.append(a+b)
```

```python
L[:5]
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()     #hamne un sabki probability nikaldi
```

```python
s
```

```python
s.plot(kind= "bar")
```

## 1.a. Cummulative density function(CDF) of PMF:

**For rolling 2 dices simultaneously:**

```python
L = []                        #empty list jisme following result store krna hai
for i in range(1000):
    a = random.randint(1,6) 
    b = random.randint(1,6) 
    L.append(a+b)
```

```python
L[:5]
```

```python
s = (pd.Series(L).value_counts()/pd.Series(L).value_counts().sum()).sort_index()     #hamne un sabki probability nikaldi
```

```python
s
```

```python
a = np.cumsum(s)    #Cummulative density function (CDF) of s which is PMF
```

```python
a
```

```python
a.plot(kind= "bar")
```
