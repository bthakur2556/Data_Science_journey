```python
#1. 3d scatterplot:

import matplotlib.pyplot as plt

fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.scatter3D(batters['runs'], batters['avg'], batters['strike_rate'], color='blue', s=30)

ax.set_xlabel("Runs")
ax.set_ylabel("Average")
ax.set_zlabel("Strike Rate")
ax.set_title("IPL Batsman analysis")

plt.show()
```

```python
#2. 3d line plot: 3D line plot is just line connecting the dots which is produced in scatterplot

x = [0,5,3]
y = [1,4,6]
z = [2,6,8]

fig = plt.figure()
ax = plt.subplot(projection='3d')

ax.scatter3D(x, y, z, s= [100,100,100])  # here s is size in x, y and z axis
ax.plot3D(x,y,z, color ="red")
```

```python
#3. 3D Surface plot (very imp in machine learning):

x = np.linspace(-10,10,100)    #line mein -10 se 10 ke beech mein 100 points dedo hamein
y = np.linspace(-10,10,100)    # yahhaan bhi line mein -10 se 10 ke beech mein 100 points dedo hamein

xx, yy = np.meshgrid(x,y) 
#xx.shape
yy.shape
```

```python
z = xx**2 + yy**2      #z = x^2 +y^2
z.shape
```

```python
fig = plt.figure(figsize = (12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z, cmap = "viridis")
fig.colorbar(p)
```

```python
#4. contour plot: it represents 3D surface plot in 2D

fig = plt.figure(figsize = (12,8))
ax = plt.subplot()   #projection='3d' hat jaayega as hamein 2d chahiye

p = ax.contour(xx,yy,z, cmap = "viridis")   #contourreplaces plot_surface
fig.colorbar(p)
```
