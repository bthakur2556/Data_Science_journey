```python
#Scatterplot:

study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9] 
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90] 
plt.scatter(study_hours, exam_scores) 
plt.title('Study Hours vs Exam Score') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Adding Color and Size: Size of points based on score (bigger score -> bigger point)

sizes = [score*2 for score in exam_scores]  #list comprehension
colors = ["red" if score<60 else "blue" for score in exam_scores]  #list comprehension
plt.scatter(study_hours, exam_scores, s=sizes, c=colors)  # s and c are built-in parameter for plt.scatter which represents size and colour respectively 

plt.title('Colored & Sized Scatter Plot') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Using a Colormap: 

import numpy as np 
scores_normalized = np.array(exam_scores) 

plt.scatter(study_hours, exam_scores, c=scores_normalized, cmap='viridis') #cmap adds gradient coloring based on the score. colorbar helps understand what the colors represent.
plt.colorbar(label='Score') 
plt.title('Scatter Plot with Colormap') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
# Adding Annotations: Annotations help identify individual points, which is useful in small datasets.

plt.scatter(study_hours, exam_scores) 

for i in range(len(study_hours)): 
    plt.annotate(f"Student {i+1}", (study_hours[i], exam_scores[i])) 
plt.title('Scatter Plot with Annotations') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Multiple Groups in One Plot: Assume two groups: Class A and Class B :

class_a_hours = [2, 4, 6, 8] 
class_a_scores = [45, 55, 65, 85] 
class_b_hours = [1, 3, 5, 7, 9] 
class_b_scores = [40, 50, 60, 70, 90] 
plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue') 
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange') 
plt.title('Scatter Plot: Class A vs Class B') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.legend() 
plt.grid(True) 
plt.show() 
```

```python
# Subplot:

import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5] 
y1 = [i * 2 for i in x]   # return double of value input
y2 = [i ** 2 for i in x]   #return square of value input

# Create a figure with 1 row and 2 columns:

plt.subplot(1, 2, 1)  # (rows, cols, plot_no), plt.subplot(1, 2, 1) means 1 row, 2 columns, and we're plotting in the 1st subplot. 
plt.plot(x, y1) 
plt.xlabel("numbers")
plt.ylabel("double of a number")
plt.title('Double of x') 
plt.subplot(1, 2, 2) #plt.subplot(1, 2, 2) means 1 row, 2 columns, and we're plotting in the 2nd subplot. 
plt.plot(x, y2) 
plt.xlabel("numbers")
plt.ylabel("square of a number")
plt.title('Square of x') 
plt.tight_layout() 
plt.show() 
```

```python
# 2×2 Grid of Subplots:This lays out 4 plots in a 2x2 grid.
 
y3 = [i ** 0.5 for i in x] 
y4 = [10 - i for i in x] 
plt.figure(figsize=(8, 6))  # Optional: make it bigger or you can use (6,6)
plt.subplot(2, 2, 1) 
plt.plot(x, y1) 
plt.title('x * 2') 
plt.subplot(2, 2, 2) 
plt.plot(x, y2) 
plt.title('x squared') 
plt.subplot(2, 2, 3) 
plt.plot(x, y3) 
plt.title('sqrt(x)') 
plt.subplot(2, 2, 4) 
plt.plot(x, y4) 
plt.title('10 - x') 
plt.tight_layout() # plt.tight_layout() avoids overlapp
plt.show() 
```
