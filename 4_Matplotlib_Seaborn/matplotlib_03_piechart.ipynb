```python
import matplotlib.pyplot as plt
# Data 
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"] 
runs = [18000, 8000, 12000, 9500] 
plt.title("Career Runs of Indian Batsmen") 
plt.pie(runs, labels = players) #we have to label that pie chart so labels= "" is must 
plt.show() 
```

```python
#Custom Colors: You can use color names or hex codes which n google {check it}.
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99'] 
plt.pie(runs, labels=players, colors=colors) 
plt.show() 
```

```python
#Add Edges and Style Slices with wedgeprops: You can customize the look of the slices using wedgeprops as it is dictionary.

plt.pie( runs, labels=players, wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '-.'}) #here i remove (colors = colors) thats why vo original colors mein vaapis aagya

plt.show()
```

```python
#Highlight a Slice Using explode: Use explode to pull out one or more slices.

explode = [0.1, 0, 0, 0]  # Only highlight Sachin's slice in order as sachin is first in list
plt.pie(runs, labels=players, explode=explode, colors=colors) #here i add (colors = colors) again, thats why vo colors mein vaapis aagya

plt.title("Exploded Pie Example") 
plt.show() 
```

```python
#if you want slice virat out even more than sachin just change variable of virat in list even more (foe example: 0.4)

explode = [0.1, 0, 0.4, 0] 
plt.pie(runs, labels=players, explode=explode, colors=colors) 
plt.title("More Exploded Pie Example") 
plt.show() 
```

```python
#Add Shadows for a 3D Feel:

plt.pie( runs, labels=players, explode=explode, colors=colors, shadow=True )# adds a 3D-like shadow 
plt.title("3D PIECHART")
plt.show() 
```

```python
#Show Percentages with autopct:

import matplotlib.pyplot as plt
activities = ["studing", "Playing", "Watching TV", "Sleeping"]
hours = [3, 2, 3, 7]
colors = ["blue", "red", "orange", "green"]
plt.figure(figsize=(6,6))  # it means all of pie chart occupy complete space
plt.pie(hours, labels = activities, colors = colors, autopct ="%1.1f%%")
plt.legend(loc = "upper left")
plt.title("student distribution")
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
# Above pie chart shows how a single student spends time in one day. Now, let’s imagine you surveyed multiple days, and you want to track how the time spent on each activity changes over a week.
#for that we use stackplot instead of pie chart


import matplotlib.pyplot as plt 
days = [1, 2, 3, 4, 5, 6, 7]  # Days of the week 
studying = [3, 4, 3, 5, 4, 3, 4]  # hours per day
playing = [2, 2, 1, 1, 2, 3, 2] 
watching_tv = [2, 1, 2, 2, 1, 1, 1] 
sleeping = [5, 5, 6, 5, 6, 5, 5] 
activities = ['Studying', 'Playing', 'Watching TV', 'Sleeping'] 
colors = ['skyblue', 'lightgreen', 'gold', 'lightcoral'] 
plt.figure(figsize=(6,6)) 
plt.stackplot(days, studying, playing, watching_tv, sleeping, labels=activities, colors=colors, alpha= 0.9) #alpha: Transparency level (0 to 1)
plt.legend(loc='upper right')  # Location of the legend 
plt.title('Weekly Activity Tracker') 
plt.xlabel('Day') 
plt.ylabel('Hours') 
plt.grid(True) 
plt.show() 
```
