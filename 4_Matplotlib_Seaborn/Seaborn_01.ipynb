```python
import seaborn as sns 
import matplotlib.pyplot as plt
import plotly.express as px     #this library is not for plotting graph but they have some datasets in which we work on
```

```python
tips = sns.load_dataset("tips") #load built-in dataset from seaborn library
tips
```

```python
#1.Relational plot:
#scatterplot: it is axes level function

sns.scatterplot(data=tips, x="total_bill", y="tip", hue="sex", style="time", size="size")  #both tip and total_bill are numerical value not catoorical value thats ehy scatterplot bana paaye 
#x= and y= replaces plt.xlabel and plt.ylabel and hue replaces plt.legend()
```

```python
#b. relplot:it is figure level function on which axes level function like above scatterplot is built 

"""seaborn  bhi ise use krne ko suggest krte hai, rather that  scatterplot. Also kyunki is relplot se hum aage axes level function bana skte hain,
just change tyhe value of 'kind'. Also relplot mein legend outside the graph hota hai kyunki yeh figure level function hai ,
so iske paas poora control hai whereas if you use lineplot, to legend graph ke andar hoga as they are axes level function to sirf graph tak access 
hai unke paas"""


sns.relplot(data= tips, x ="total_bill", y="tip", kind="scatter", hue="sex", style="time", size="size")
```

```python
gap = px.data.gapminder()
gap
```

```python
data1 = gap[gap["country"] == "India"]  # filter the dataset and set only for india
data1
```

```python
#c. lineplot: it is also axes level function

sns.lineplot(data = data1, x="year", y="lifeExp")
```

```python
#same line plot but using relplot: only chnage value of kind

sns.relplot(data = data1, x="year", y="lifeExp", kind="line")
```

```python
temp = gap[gap["country"].isin(["China", "India", "Pakistan"])]
temp
```

```python
sns.relplot(data= temp, x="year", y="lifeExp", kind="line", hue="country")
```

```python
sns.relplot(data= tips, x="total_bill", y="tip", kind="scatter", col ="sex", row="smoker")

#row="smoker" means 2 rows banegi if they smoker or not and col= "sex" means 2 columns bnege on the basis of sex i.e. male and female 
#so, total 2X2 ki grid bnegi
```

```python
sns.scatterplot(data= tips, x="total_bill", y="tip", col ="sex", row="smoker")
```

```python
#2. Distribution plot:

#a.i. Histogram: plotting univariant histogram: axes level function

sns.histplot(data= tips, x="total_bill", bins=20)
```

```python
#b. Displot: its figure level function of distribution plot

sns.displot(data= tips, x="total_bill", kind ="hist", bins=20)
```

```python
sns.displot(data= tips, x="total_bill", kind ="hist", bins=20, hue="sex")
```

```python
#faceting using row and col: but remember yeh sirf figure level function par use hoga naaki axes level pe

sns.displot(data= tips, x="total_bill", kind ="hist", bins=20, col="sex")
```

```python
#c. kdeplot: rather than discrete graphs they show continous graph for distribution plot which they mention absent values also in above plots

sns.kdeplot(data= tips, x="total_bill")
```

```python
#kdeplot using displot:

sns.displot(data= tips, x="total_bill", kind = "kde")
```

```python
#a.ii. Bivariate histogram: 2 parameters jahaan collide hue vahan mil jaayenge

sns.histplot(data= tips, x ="total_bill", y= "tip")
```

```python
#3. matrix plot:

#a. Heatmap:

gap
```

```python
#above table is too long for heatmap to cover so we are going to wide table:

temp1= gap.pivot(index= "country",columns="year", values="lifeExp")
temp1
```

```python
sns.heatmap(temp1)
```

```python
#if we need only european countries: to pehle sirf wide table banate hai:

temp2 = gap[gap["continent"] == "Europe"].pivot(index= "country",columns="year", values="lifeExp")
temp2
```

```python
plt.figure(figsize = (10,8))
sns.heatmap(temp2)
```

```python
#if we want to label it like what value each box contain : use annotation i.e. annot function

plt.figure(figsize = (10,8))
sns.heatmap(temp2, annot = True, linewidth = 0.5)  #linewidth for better visibility like it creates gap between boxes
```
