```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
plt.bar(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
#Setting Bar Width & Side-by-Side Bar Charts Let’s compare Sachin, Sehwag, and Kohli side-by-side for the same years.

import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))
width = 0.25
plt.bar(x - width, sachin, color="blue", width = width, label = "Sachin") 
plt.bar(x, sehwag, color="red", width = width, label = "Sehwag") 
plt.bar(x + width, kohli, color="black", width = width, label = "Virat") 
plt.xlabel("Year")
plt.ylabel("Runs scored")
plt.title("yearly run comparison")
plt.legend()
plt.show()
```

```python
#now as you can see that above bar graph, x axis is not the years that we mentioned earlier. its just indexes so to remove that we just add one line rest of code is same

import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years)) # np.arange() Helps to align multiple bars side by side
width = 0.25

plt.bar(x - width, sachin, width=width, color="blue", label = "Sachin")
plt.bar(x, sehwag, width=width, color="red", label = "Sehwag")
plt.bar(x + width, kohli, width=width, color="black", label = "Kohli")
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Yearly Run Comparison")
plt.xticks(x, years)  # Replaces index numbers with real labels: now it Show actual year instead of 0,1,2,...:
plt.legend()
plt.tight_layout()
```

```python
#Horizontal Bar Charts with barh(): Let’s compare total runs scored in the first 5 years by different players.

players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [500+700+1100+1500+1800, 0+200+900+1400+1600, 0+0+500+800+1100, 300+600+800+1100+900]
plt.barh(players, runs_5yrs, color="skyblue")
plt.xlabel("Total Runs in First 5 Years")
plt.ylabel("Players")
plt.title("First 5-Year Performance of Indian Batsmen")
plt.tight_layout()  #Prevents labels from overlapping
plt.show()
```

```python
#Adding Value Labels with plt.text(): You can use plt.text() to display values on top of bars, making your chart easier to read. It's especially useful in presentations and reports.

import matplotlib.pyplot as plt
 
players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]
 
plt.bar(players, runs, color="red")
 
for i in range(len(players)): 
    plt.text(i, runs[i] + 20,str(runs[i]), ha='center') #Use ha='center' to center the text, str(runs[i]) is text taht we just print on bar so it must be strings and add a small offset (+20 here) to avoid overlap with the bar.
plt.ylabel("Runs")
plt.title("Runs Scored by Players")
plt.tight_layout()
plt.show()
```
