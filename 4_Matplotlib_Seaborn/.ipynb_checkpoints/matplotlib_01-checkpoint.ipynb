```python
import matplotlib.pyplot as plt
years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010] 
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500] 
plt.plot(years,runs)
```

```python
#Above graph looks discrete so make it more easily readable graph

plt.plot(years, runs)
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("Sachin Tendular's Scorecard")
plt.show()
```

```python
#Multiple Lines in One Plot

kohli = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100] 
sehwag = [0, 300, 800, 1200, 1500, 1700, 1600, 1400, 1000, 0] 

plt.plot(years, kohli, label="Virat Kohli")  #here kohli is y-axis values
plt.plot(years, sehwag, label="Virendra Sehwag") #here sehwag is y-axis values
plt.xlabel("Years")
plt.ylabel("Runs")
plt.title("2 Batsman Scorecard")
plt.legend() #to show which colour represents which player
plt.show()
```

```python
#Color, Line Style and line width Arguments:

plt.plot(years, kohli, color="orange", linestyle= '--',linewidth=4, label = "Virat Kohli")  #line is thickest than rest of others
plt.plot(years, sehwag, color="blue", linestyle= '-.',linewidth=3, label = "Viru Sehwag")   #line is thinner than virat kohli but thicker than sachin
plt.plot(years, runs, color="black",linewidth=2, label = "Sachin Tendulkar")  #line is thinnest 
plt.legend()
```

```python
#Layout Tweaks: using grid(true) and tight_layout() function

plt.plot(years, kohli, linewidth=3, label="Kohli") 
plt.plot(years, sehwag, linewidth=2, label="Sehwag") 
plt.legend() 
plt.grid(True)  #to make grid
plt.tight_layout() # to make graph look perfect fit
plt.show() 
```

```python
#Visualizing Tons of Data – What Crowded Looks Like

import numpy as np
for i in range(50):
    plt.plot(np.random.rand(100), linewidth = 1)

plt.title("Too much data to handle")
plt.grid(True)
plt.tight_layout()
plt.show()
```
