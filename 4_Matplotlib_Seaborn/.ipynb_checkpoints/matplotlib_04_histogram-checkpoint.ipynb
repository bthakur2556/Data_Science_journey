```python
# Histogram:1. bins
# plt.hist(data, bins=10)  # 10 equal-width bins 
# plt.hist(data, bins=[10, 20, 30, 40, 60, 100])  # Custom age bins

import matplotlib.pyplot as plt 
ages = [ 
34, 28, 36, 45, 27, 27, 45, 37, 25, 35, 
25, 25, 32, 10, 12, 24, 19, 33, 20, 15, 
44, 27, 30, 15, 24, 31, 18, 33, 23, 27, 
23, 48, 29, 19, 38, 17, 32, 10, 16, 31, 
37, 31, 28, 26, 15, 22, 25, 40, 33, 12, 
33, 26, 23, 36, 40, 39, 21, 26, 33, 39, 
25, 28, 18, 18, 38, 43, 29, 40, 33, 23, 
33, 45, 29, 45, 3, 38, 30, 27, 30, 10, 
27, 33, 44, 24, 21, 24, 39, 33, 24, 35, 
30, 39, 22, 26, 26, 15, 32, 32, 30, 27 
] 
gaps = [10, 20, 30, 40, 50, 60, 70] 
plt.hist(ages, bins=gaps) 
plt.title('Age Distribution of YouTube Viewers') 
plt.xlabel('Age Group') 
plt.ylabel('Number of Viewers') 
plt.grid(True, linestyle='--', alpha=0.5) 
plt.show() 
```

```python
#add eggdecolor to make border of each bar: edgecolor for Better Visibility, this parameter adds borders to the bars, improving clarity.

plt.hist(ages, bins=gaps, edgecolor='black') 
plt.title('Age Distribution of YouTube Viewers') 
plt.xlabel('Age Group') 
plt.ylabel('Number of Viewers') 
plt.grid(True, linestyle='--', alpha=0.5) 
plt.show() 
```

```python
#Adding a Vertical Line: axvline: Use axvline to mark a specific value like the average age or a threshold.

import numpy as np
plt.hist(ages, bins=gaps, edgecolor='black') 
plt.axvline(np.mean(ages), color='red', linestyle='--', linewidth=2, label='Average Age') 
plt.legend() 
plt.title('Age Distribution with Mean Line') 
plt.show() 
```

```python
#Scatterplot:

study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9] 
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90] 
plt.scatter(study_hours, exam_scores) 
plt.title('Study Hours vs Exam Score') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Adding Color and Size: Size of points based on score (bigger score -> bigger point)

sizes = [score*2 for score in exam_scores]
colors = ["red" if score<60 else "blue" for score in exam_scores]
plt.scatter(study_hours, exam_scores, s=sizes, c=colors)

plt.title('Colored & Sized Scatter Plot') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Using a Colormap: 

import numpy as np 
scores_normalized = np.array(exam_scores) 

plt.scatter(study_hours, exam_scores, c=scores_normalized, cmap='viridis') #cmap adds gradient coloring based on the score. colorbar helps understand what the colors represent.
plt.colorbar(label='Score') 
plt.title('Scatter Plot with Colormap') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
# Adding Annotations: Annotations help identify individual points, which is useful in small datasets.

plt.scatter(study_hours, exam_scores) 

for i in range(len(study_hours)): 
    plt.annotate(f'Student {i+1}', (study_hours[i], exam_scores[i])) 
plt.title('Scatter Plot with Annotations') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.grid(True) 
plt.show() 
```

```python
#Multiple Groups in One Plot: Assume two groups: Class A and Class B :

class_a_hours = [2, 4, 6, 8] 
class_a_scores = [45, 55, 65, 85] 
class_b_hours = [1, 3, 5, 7, 9] 
class_b_scores = [40, 50, 60, 70, 90] 
plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue') 
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange') 
plt.title('Scatter Plot: Class A vs Class B') 
plt.xlabel('Study Hours') 
plt.ylabel('Exam Score') 
plt.legend() 
plt.grid(True) 
plt.show() 
```

```python
# Subplot:

import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5] 
y1 = [i * 2 for i in x] 
y2 = [i ** 2 for i in x] 

# Create a figure with 1 row and 2 columns:

plt.subplot(1, 2, 1)  # (rows, cols, plot_no), plt.subplot(1, 2, 1) means 1 row, 2 columns, and we're plotting in the 1st subplot. 
plt.plot(x, y1) 
plt.xlabel("numbers")
plt.ylabel("double of a number")
plt.title('Double of x') 
plt.subplot(1, 2, 2) #plt.subplot(1, 2, 2) means 1 row, 2 columns, and we're plotting in the 2nd subplot. 
plt.plot(x, y2) 
plt.xlabel("numbers")
plt.ylabel("square of a number")
plt.title('Square of x') 
plt.tight_layout() 
plt.show() 
```

```python
# 2×2 Grid of Subplots:This lays out 4 plots in a 2x2 grid.
 
y3 = [i ** 0.5 for i in x] 
y4 = [10 - i for i in x] 
plt.figure(figsize=(8, 6))  # Optional: make it bigger or you can use (6,6)
plt.subplot(2, 2, 1) 
plt.plot(x, y1) 
plt.title('x * 2') 
plt.subplot(2, 2, 2) 
plt.plot(x, y2) 
plt.title('x squared') 
plt.subplot(2, 2, 3) 
plt.plot(x, y3) 
plt.title('sqrt(x)') 
plt.subplot(2, 2, 4) 
plt.plot(x, y4) 
plt.title('10 - x') 
plt.tight_layout() # plt.tight_layout() avoids overlapp
plt.show() 
```

```python
#Now lets look into some plots we can create usign Seaborn: 
#1. Line Plot

tips = sns.load_dataset("tips")
sns.lineplot(x = "tip", y = "total_bill", data = tips) # we have to tell them that which data we are comsidering, so in this case data = tips
plt.title('Line Plot Example') 
plt.show() 
```

```python
#2. Scatterplot:

sns.scatterplot(x="total_bill", y="tip", data=tips, hue="time") # hue parameter is used to add a categorical dimension to your plots by color-coding data points or bars based on the values of another variable
plt.title('Scatter Plot with Color by Time') 
plt.show() 
```

```python
#3. barplot:

sns.barplot(x="day", y="total_bill", data=tips) 
plt.title('Average Bill per Day') 
plt.show() 
```

```python
#boxplot:

sns.boxplot(x="day", y="total_bill", data=tips) 
plt.title('Boxplot of Total Bill per Day') 
plt.show() 
```
