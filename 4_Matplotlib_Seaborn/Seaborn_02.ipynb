```python
import seaborn as sns 
import matplotlib.pyplot as plt
import plotly.express as px     #this library is not for plotting graph but they have some datasets in which we work on
```

```python
tips = sns.load_dataset("../datasets/tips")
iris = sns.load_dataset("../datasets/iris")
```

#1.a. stripplot : Axes level function:

#as scatterp[lot mein dono x and y values numerical thi, isme ek value catogorical ho skti hai

```python
sns.stripplot(data= tips, x="day", y="total_bill")  #here days are categorical and total_bills are numerical
```

#b. stripplot using catplot as catplot is figure level function of this

```python
sns.catplot(data= tips, x="day", y="total_bill", kind="strip", hue="sex")
```

```python
#2.a. Boxplot:

sns.boxplot(data= tips, x="day", y="total_bill")  #day 4 hai isliye 4 boxplot bane, if x= "sex" hota:to 2 box plot bnte, 1: male and 1:female
```

```python
sns.boxplot(data= tips, x="day", y="total_bill", hue="sex")  #8 : 4 for each sex
```

```python
#b. Violin plot: (kde plot + box plot):

sns.violinplot(data= tips, x="day", y="total_bill")
```

```python
#3.a. Barplot:

sns.barplot(data= tips, x="day", y="total_bill", hue="smoker")
```
