```python
import pandas as pd
df = pd.DataFrame()  #empty dataframe jiske andar hum apni  files daalenge
```

**a. export as to_csv:**

```python
df1 = pd.read_csv("../datasets/batter.csv")
df1.head()
```

```python
df1.groupby("batter")["runs"].sum()
```

```python
#above is series right now , so convert in dataframe:

temp = df1.groupby("batter")["runs"].sum().reset_index()
temp
```

```python
#now export above table in new csv file:

temp.to_csv("batsman_run.csv", index = False)   #ab ek new csv file generate ho gyi batsman_run.csv naam se
```

**b. export as to_excel:**

```python
df1.groupby("batter")["runs"].sum().reset_index()
```

```python
#now export above table in new excel file:

temp.to_csv("batsman_run.xlsx", index = False)   #ab ek new csv file generate ho gyi batsman_run.xlsx naam se
```

**c. export as to json:**

```python
df2 = df1.groupby(["batter", "runs"])["strike_rate"].sum().unstack()
df2
```

```python
df2.to_json("batsmen.json")  #nnow export to json
```

```python
import pymysql
from sqlalchemy import create_engine   # yeh 2 libraries bahut imp hain sql export mein
```

```python
engine = create_engine("mysql+pymysql://root:@localhost/batters") #fir aapko ek engine object banana hai

#syntax: //(root):(password)@(url)/(database)  here: password is bothing isliye kuch nhi like and database hamne batters ke naam se banaya hai
```

```python
df1.to_sql("Batsman_data", con = engine, if_exists = "append") #now if you see in localhost page, tum dekhoge ki vahaan batters databases mein Batsman_data naam ki sql file add ho gyi hai
```

```python
temp.to_sql("Batsman_run", con = engine, if_exists = "append") 
df2.to_sql("unstack_batsman", con = engine, if_exists = "append") 
```
