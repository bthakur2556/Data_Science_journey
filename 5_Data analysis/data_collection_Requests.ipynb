```python
#so for that we just have to do:

for i in range(1,51):
    a = requests.get(f"https://books.toscrape.com/catalogue/page-{i}.html")  #here page-1 change to page-{i} as 1-50 tak jaana hai
    with open(f"htmls/page{i}.html", "w", encoding = "utf-8") as f:
        f.write(a.text)
        print(f"Successfully Downlaod page {i}") 
```

```python
#Using Beautiful Soup for Data Collection:

from bs4 import BeautifulSoup
```

```python
with open("htmls/page1.html") as f:
    content =  f.read()
```

```python
soup = BeautifulSoup(content, "html.parser")
```

```python
#fir hum us link pe jayenge jisko humne page1.html mein daala hai i.e "https://books.toscrape.com/catalogue/page-1.html" and when we right click on title of first book and click inspect
#then we got to know that titles, product price and rating(that we needed), all are under article named "product_pod" and title specifically under h3 header

articles = soup.select("article.product_pod")  #as all are under article named "product_pod"
h3s = soup.find_all("h3") # as all titles are specifically under h3 header to vo hamein find krna padega
```

```python
#ab article ke andar vo saari cheezein hai jo hamein chahiye like titles, product price and rating and hamein yeh saara kuch print krna hai so why not make alist which includes these things so:

items = []
for article in articles:
    title = article.find("h3").find("a")["title"]   # as title is under h3 header and uske andar bhi it is in "Anchor tag <a>" 
    price = article.select_one("p.price_color").text.split("£")[1] #as price is under "<p> Paragraph text" i.e. price_color and hamein text form mein chahiye vo bhi and if we run just upto text to vo output hamein "Ã‚Â£51.77" aisa de rha hai so thats why we soplit from "£" symbol  
    rating_element = article.select_one("p.star-rating") #as first book liye to direct star-rating Three likha tha but andar classes thi jo teen active thi so andar ke classes ko bhi active krna tha taaki baaki books ke automatically book rating mil jaye
    rating = rating_element["class"][1]
    items.append([title,price,rating])  # final list mein teeno cheeze aagyi
    
```

```python
import pandas as pd
df = pd.DataFrame(items, columns = ["Book Title", "Price (in £)", "Rating (out of five)"])
```
