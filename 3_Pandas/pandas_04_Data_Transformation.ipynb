```python
import pandas as pd
df = pd.read_csv("data.csv")
```

```python
df
```

```python
df.sort_values("Year") #automatically ascending order mein change ho jayega
```

```python
df.sort_values("Year", ascending = False) #ab descending mein change ho jayega and note that indexees bhi shift ho chuke hain
```

```python
df.sort_values(["Year", "IMDb"]) 
#df.sort_values([“Year”, “IMDb”]) sorts the DataFrame first by the “Year” column, andif there are ties (i.e., two or more rows with the same “Year”), it will sort by the“IMDb” column
```

```python
#If you want the index to start from 0 and be sequential, you can reset it using reset_index()
df.reset_index(drop = True)
```

```python
df.reset_index(drop=True, inplace=True) # If you want to Reset the index and drop the old index (like copy nhi direct view hi chnge krna hai to inplace= True)krde
```

```python
df #Ab original table bhi change ho gya "inplace = True" ki help se
```

```python
df["Rank"] = df["IMDb"].rank(ascending = False)  #it provides new column named "Rank" jo highest number/rank to highest IMDb dega
```

```python
df.sort_values("Rank")
```
