```python
df.value_counts()
```

```python
#Filtering with Conditions: 
#a. Simple Condition
df[df["Year"] > 2018]
```

```python
#b. Multiple Conditions (AND / OR)
df[(df["Year"]>2018) & (df["IMDb"]>7)] #and operator 
```

```python
df[(df["Year"]>2018) | (df["IMDb"]>7)] #or operator
```

```python
#Querying with .query(): it makes more convinient code
df.query("Year> 2018 and IMDb> 7")  #here "and operator(&)" is replaced with "and"
```

```python
df.query("Year>2018 or IMDb>7")    #here "or operator(|)" is replaced with "or"
```
