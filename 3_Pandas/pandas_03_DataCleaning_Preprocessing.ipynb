```python
df["Age"] = df["Age"].astype(int) #as age is float64 but we need integer so typecasted
```

**IntCastingNaNError: Cannot convert df table, as non-finite values (NA or inf) are present in df table,especially age column kyunki age ka nikalna tha but vahan NAN value thi, which cannot typecasted into integer. 
so df1=df.dropna() se jo df1 table baaaya tha vo use krenge as usmein hamne sari missing values and NaN values reove krdi this
and also we make a copy out of it**

```python
df1["Age"] = df1["Age"].astype(int).copy() #as humne pehle df1=df.dropna() # Drop rows with *any* missing values or NaN values
```

```python
df1
```

```python
df1.info()  #now in "age" column, datatype of df1 table is changed from "float64" to "int 64"
```

```python
#Applying Functions: 

#1. .apply() → Apply any function to rows or columns

df1["Age Group"] = df1["Age"].apply(lambda x: "Adult" if x >= 25 else "Minor").copy()
```

```python
df1 #new column with age group
```

```python
#2. .map() → Element-wise mapping for Series
gender_map = {"M": "Male", "F": "Female"}
df1["Gender"] = df1["Gender"].map(gender_map).copy()
```

```python
df1
```

```python
#3. .replace() → Replace specific values
df1["City"] = df1["City"].replace({"New York": "New Yorkery", "Mumbai": "New Mumbai"})
```
