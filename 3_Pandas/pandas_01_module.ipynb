```python
#1. core data structures in pandas: 

#a. Series — 1D Labeled Array: A Series is like a list with labels (index)
import pandas as pd
s = pd.Series([10, 20, 30, 40, 50])
print(s)
```

```python
s = pd.Series([10, 20, 30, 40, 50], index= ["Bhanu", "Aditya", "harry", "Swati", "dhruv"]) #You can also define custom index
print(s)
```

```python
#b. DataFrame — 2D Labeled Table: A DataFrame is like a dictionary of Series — multiple columns with labels.

data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "city": ["Delhi", "Mumbai", "Bangalore"]
}
 
df = pd.DataFrame(data)
```

```python
df
```

```python
df.index = ["a", "b", "c"]  #changes default indexes to a, b and c 
df.columns = ["Name", "Age", "City"] #changes columns with changing capital letter
```

```python
df
```

```python
#2. creating dataframes from different method:

#1. From Python Lists:

import pandas as pd
 
data = [
    ["Alice", 25],      #these are list of lists
    ["Bob", 30],
    ["Charlie", 35]
]
 
df = pd.DataFrame(data, columns= ["Names","Age"])
```

```python
df
```

```python
#2. From Dictionary of Lists:

data = {
    "Name" : ["Alice", "Bob", "Charlie"],
    "Age" : [30, 40, 50]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
#3. From NumPy Arrays:

import numpy as np
arr = np.array([[1,2], [3,4]])
df = pd.DataFrame(arr, columns=["A", "B"])
```

```python
df
```
