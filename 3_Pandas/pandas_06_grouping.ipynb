```python
#Grouping and aggregating helps you summarize your data — like answering:
#“What’s the average salary per department?”
#“How many users joined the Gym per month?”
```

```python
import pandas as pd
data = ({
    "Department": ["HR", "HR", "IT", "IT", "Marketing", "Marketing", "Sales", "Sales"],
    "Team": ["A", "A", "B", "B", "C", "C", "D", "D"],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F"],
    "Salary": [85000, 90000, 78000, 85000, 92000, 88000, 75000, 80000],
    "Age": [23, 25, 30, 22, 28, 26, 21, 27],
    "JoinDate": pd.to_datetime([
    "2020-01-10", "2020-02-15", "2021-03-20", "2021-04-10",
    "2020-05-30", "2020-06-25", "2021-07-15", "2021-08-01"
    ])
})

df = pd.DataFrame(data)
```

```python
df
```

```python
#“Group by Department, then calculate average Salary for each group.”
#.groupby: for grouping

df.groupby("Department")["Salary"].mean()  # Average per team
```

```python
#To group by multiple columns:

df.groupby(["Team", "Gender"])["Salary"].mean()
```

```python
#custom operations:

#1. Custom Aggregations with .agg(): Apply multiple functions at once like this:
df.groupby("Team")["Salary"].agg(["mean", "max", "min"])
```

```python
#Name your own functions:

df.groupby("Team")["Salary"].agg(
    avg_salary="mean",
    highest_salary="max",
    lowest_salary = "min"
)
```

```python
#Apply different functions to different columns:

df.groupby("Team").agg({
"Salary": "mean",
"Age": "max"
})
```

```python
#2. .transform() Example:

df["Team Avg"] = df.groupby("Team")["Salary"].transform("mean") #add a new column named "Team Avg" jo "salary" ka mean nikalega by grouping "team" 
```

```python
df  #Now each row gets its team average — great for comparisons!
```

```python
#3. .filter() Example:

df.groupby("Team").filter(lambda x: x["Salary"].mean() > 85000)  #Only keeps teams with average salary > 85000.
```

```python
# Merging & Joining Data: Often, data is split across multiple tables or files. Pandas lets you combine them just like SQL — or even more flexibly

employees = pd.DataFrame({
    "EmpID": [1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "DeptID": [10, 20, 30]
})
departments = pd.DataFrame({
    "DeptID": [10, 20, 40],
    "DeptName": ["HR", "Engineering", "Marketing"]    
})
```

```python
employees
```

```python
departments
```

```python
#Merge Like SQL: pd.merge()

pd.merge(employees, departments, on="DeptID") #Inner Join (default)
#Returns only matching DeptIDs
```

```python
pd.merge(employees, departments, on ="DeptID", how= "left") #left join: Keeps all employees, fills NaN where no match.
```

```python
pd.merge(employees, departments, on ="DeptID", how= "right") #right join: Keeps all departments, even if no employee.
```

```python
pd.merge(employees, departments, on="DeptID", how="outer")  #outer join: Includes all data, fills missing with NaN .
```

```python
#looping in groups:

#name highest rated movie in each genre:

for group,data in genres:
    print(data[data["IMDB_Rating"] == data["IMDB_Rating"].max()]) 
```

```python
#now you got highest rated movie in each genre, but not clear so add empty list for making table then usme yeh append krdenge

result = []

for group, data in genres:
    result.append(data[data["IMDB_Rating"] == data["IMDB_Rating"].max()])

df = pd.concat(result, ignore_index=True)
```

```python
#using apply function:

#fing no of movies starting from A from each genre:

def abc(group):               #make custom function named abc who take parameter a group and return it
    return group["Series_Title"].str.startswith("A").sum()    #humne ".str" is liye lagaya kyunki group ka type dataframe hai to string mein change krna padega and ".sum" isliye kiya kyunki vran vo boolean result deta agar hum .sum nhi lagate  
```

```python
genres.apply(abc)
```

```python
#find normalized IMDB rating groupwise:

def normal(group):
    group["norm_rating"] = group["IMDB_Rating"] - group["IMDB_Rating"].min()/ (group["IMDB_Rating"].max()) - (group["IMDB_Rating"].min())
    return group
```

```python
genres.apply(normal) #naya column named norm_rating add ho gya
```
