```python
#1 find the memory size of a NumPy array with 1 million elements.

import numpy as np 
import time
import sys
```

```python
list1 = list(range(1000000))
np1 = np.array(list1)
a = np1.nbytes
print(f"numpy takes {a} bytes") 
```

```python
#2 compare the time taken to multiply two Python lists vs. two NumPy arrays.
#a python list:

list1 = list(range(100000))
list2 = list(range(100000))
start = time.time()
result = [x*y for x,y in zip(list1,list2)]
end = time.time()
print("time taken by this operation to run sucessfully is", end - start)
```

```python
#b. numpy arrays:

arr1 = np.array(list1)
arr2 = np.array(list2)
start =  time.time()
result = arr1*arr2
end = time.time()
print("time taken by this operation to run sucessfully is", end - start)
```
