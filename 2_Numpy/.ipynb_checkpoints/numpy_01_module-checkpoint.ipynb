```python
#using python list 
import numpy as np 
import time

size = 1_000_000
l1 = list(range(size))
l2 = list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()

print(end - start)
```

```python
#using numpy array
import numpy as np 
import time

size = 1_000_000
#making numpy array using list
list1 = np.array(list(range(size)))
list2 = np.array(list(range(size)))

start = time.time()
add = l1 + l2
end = time.time()

print(end - start)
```

```python
#now chech memory it takes, so import "sys" module
import sys
list_data = list(range(1000))
numpy_data = np.array(list_data)
print(f"python list takes: {(sys.getsizeof(list_data)*len(list_data))} bytes")
print(f"numpy array takes: {numpy_data.nbytes} bytes")

```
